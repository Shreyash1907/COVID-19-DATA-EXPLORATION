```python
import pandas as pd
import numpy as np

data = {
    'Country': ['India', 'USA', 'Brazil', 'UK', 'Russia', 'Japan', 'Germany', 'France'],
    'Confirmed': [45000000, 103000000, 37000000, 25000000, 23000000, 33000000, 39000000, 40000000],
    'Deaths': [533000, 1200000, 700000, 230000, 400000, 74000, 180000, 170000],
    'Recovered': [44500000, 101000000, 36000000, 24500000, 22000000, 32000000, 38500000, 39500000]
}

df = pd.DataFrame(data)

print("COVID-19 DATASET")
print(df)

print("\nDATASET INFORMATION")
df.info()

print("\nMISSING VALUES")
print(df.isnull().sum())

print("\nSTATISTICAL SUMMARY")
print(df.describe())
```

```python
df['Active_Cases'] = df['Confirmed'] - df['Deaths'] - df['Recovered']
df['Death_Rate'] = (df['Deaths'] / df['Confirmed']) * 100

print("ACTIVE COVID-19 CASES")
print(df[['Country', 'Active_Cases']])

print("\nCOUNTRY-WISE DEATH RATE (%)")
print(df[['Country', 'Death_Rate']].round(2))
```

```python
print("COUNTRY WITH HIGHEST CONFIRMED CASES")
print(df.loc[df['Confirmed'].idxmax(), ['Country', 'Confirmed']])

print("\nCOUNTRY WITH LOWEST CONFIRMED CASES")
print(df.loc[df['Confirmed'].idxmin(), ['Country', 'Confirmed']])

print("\nCOUNTRIES RANKED BY CONFIRMED CASES")
print(df.sort_values('Confirmed', ascending=False)[['Country', 'Confirmed']])
```

```python
print("TOTALS FOR THE SAMPLE DATASET")
print("Confirmed Cases:", df['Confirmed'].sum())
print("Deaths:", df['Deaths'].sum())
print("Recovered Cases:", df['Recovered'].sum())

print("\nCOUNTRIES WITH MORE THAN 30 MILLION CASES")
print(df[df['Confirmed'] > 30000000][['Country', 'Confirmed']])

print("\nFINAL COVID-19 DATASET")
print(df.round(2))
```
